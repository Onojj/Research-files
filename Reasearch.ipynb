```python
import pandas as pd
import numpy as np

# --- Load ---
main_path = r"C:\Users\MIKE\CHARLS H\H_CHARLS_D_Data.dta"
lh_path = r"C:\Users\MIKE\CHARLS_LH\H_CHARLS_LH_a.dta"

df_main = pd.read_stata(main_path)
df_lh = pd.read_stata(lh_path)

# --- Merge ---
df = pd.merge(df_main, df_lh, on='ID', how='inner', suffixes=('', '_lh'))
print(f"Merged shape: {df.shape}")

# --- Build childhood health adversity ---
def to_numeric_cat(s):
    """Convert Stata categorical to numeric, keeping NaN."""
    return pd.to_numeric(s.cat.codes.replace(-1, np.nan), errors='coerce')

# rachchlt: 1=Much healthier ... 5=Much less healthy
# Recode to binary: 1 if 4 or 5, else 0
rachchlt_num = to_numeric_cat(df['rachchlt'])
df['child_ill_health'] = ((rachchlt_num == 3) | (rachchlt_num == 4)).astype(float)
df.loc[rachchlt_num.isna(), 'child_ill_health'] = np.nan

# ramischlth, rachbedhlth: 0/1 already
df['child_missed_school'] = to_numeric_cat(df['ramischlth'])
df['child_bed_confined'] = to_numeric_cat(df['rachbedhlth'])

# Composite: sum of the three, only if at least 2 of 3 non-missing
composite_cols = ['child_ill_health', 'child_missed_school', 'child_bed_confined']
df['childhood_health_adversity'] = df[composite_cols].sum(axis=1, skipna=False)
df.loc[df[composite_cols].notna().sum(axis=1) < 2, 'childhood_health_adversity'] = np.nan

# --- Build childhood vaccination ---
df['childhood_vaccinated'] = to_numeric_cat(df['rachvaccine'])

# --- Build childhood SES ---
mom_occ = to_numeric_cat(df['ramomoccup_c'])  # 0=Farming, 1=Non-agri
dad_occ = to_numeric_cat(df['radadoccup_c'])
# Disadvantaged = both parents farming
df['childhood_ses_disadvantaged'] = ((mom_occ == 0) & (dad_occ == 0)).astype(float)
df.loc[mom_occ.isna() | dad_occ.isna(), 'childhood_ses_disadvantaged'] = np.nan

# --- Inspect ---
print("\n" + "=" * 70)
print("CHILDHOOD CONSTRUCT DISTRIBUTIONS")
print("=" * 70)

new_features = [
    'childhood_health_adversity',
    'childhood_vaccinated',
    'childhood_ses_disadvantaged'
]

for f in new_features:
    print(f"\n--- {f} ---")
    print(f"Non-null: {df[f].notna().sum()} / {len(df)} ({df[f].notna().mean()*100:.1f}%)")
    print(df[f].value_counts(dropna=False).sort_index())

# --- Correlation with outcome (if target can be built) ---
if 'r1cesd10' in df.columns and 'r2cesd10' in df.columns and 'r3cesd10' in df.columns:
    df['cesd_delta_w2_w3'] = df['r3cesd10'] - df['r2cesd10']
    df['y_dep_decline_w3'] = (df['cesd_delta_w2_w3'] > 0).astype(int)

    print("\n" + "=" * 70)
    print("CORRELATION WITH OUTCOME (y_dep_decline_w3)")
    print("=" * 70)

    for f in new_features:
        valid = df[[f, 'y_dep_decline_w3']].dropna()
        if len(valid) > 100:
            corr = valid[f].corr(valid['y_dep_decline_w3'])
            mean_by_class = valid.groupby('y_dep_decline_w3')[f].mean()
            print(f"\n{f}:")
            print(f"  Pearson r with outcome: {corr:.4f}")
            print(f"  Mean by outcome class:")
            print(mean_by_class)

    # --- Correlation with hacohort_c (to confirm it's a separate signal) ---
    print("\n" + "=" * 70)
    print("CORRELATION WITH hacohort_c (sampling cohort)")
    print("=" * 70)
    hacohort_num = to_numeric_cat(df['hacohort_c'])
    for f in new_features:
        valid = df[[f]].copy()
        valid['hacohort'] = hacohort_num
        valid = valid.dropna()
        if len(valid) > 100:
            corr = valid[f].corr(valid['hacohort'])
            print(f"{f} vs hacohort_c: r = {corr:.4f}")
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score, f1_score, accuracy_score
from sklearn.impute import SimpleImputer
import shap

# ==========================================
# 1. LOAD & MERGE
# ==========================================
print("Loading CHARLS datasets...")
main_path = r"C:\Users\MIKE\CHARLS H\H_CHARLS_D_Data.dta"
lh_path = r"C:\Users\MIKE\CHARLS_LH\H_CHARLS_LH_a.dta"

df_main = pd.read_stata(main_path)
df_lh = pd.read_stata(lh_path)
df = pd.merge(df_main, df_lh, on='ID', how='inner', suffixes=('', '_lh'))
print(f"Merged shape: {df.shape}")

# ==========================================
# 2. BUILD TARGET
# ==========================================
df['cesd_delta_w2_w3'] = df['r3cesd10'] - df['r2cesd10']
df['y_dep_decline_w3'] = (df['cesd_delta_w2_w3'] > 0).astype(int)

# ==========================================
# 3. BUILD Famine EXPOSURE
# ==========================================
def classify_famine(birth_year):
    if pd.isna(birth_year):
        return np.nan
    if birth_year >= 1962:
        return 0
    elif 1959 <= birth_year <= 1961:
        return 1
    elif 1956 <= birth_year <= 1958:
        return 2
    elif birth_year < 1956:
        return 3
    return np.nan

df['famine_exposure'] = df['rabyear'].apply(classify_famine)

# ==========================================
# 4. HACOHORT NUMERIC
# ==========================================
df['hacohort_c_num'] = pd.to_numeric(
    df['hacohort_c'].cat.codes.replace(-1, np.nan), errors='coerce'
)

# ==========================================
# 5. DEFINE FEATURE SETS
# ==========================================
features_a = ['r1cesd10', 'r2hosp1y']
features_b = ['r1cesd10', 'r2hosp1y', 'hacohort_c_num', 'r1agey', 'ragender', 'raeduc_c']
features_c = ['r1cesd10', 'r2hosp1y', 'famine_exposure', 'r1agey', 'ragender', 'raeduc_c']

all_features = list(set(features_a + features_b + features_c))
model_data = df[all_features + ['y_dep_decline_w3']].dropna(subset=['y_dep_decline_w3']).copy()

# Coerce non-numeric to numeric
for col in all_features:
    if model_data[col].dtype == 'object' or str(model_data[col].dtype) == 'category':
        model_data[col] = model_data[col].astype('category').cat.codes
    else:
        model_data[col] = pd.to_numeric(model_data[col], errors='coerce')

y = model_data['y_dep_decline_w3'].values
print(f"\nAnalytic N: {len(model_data)}")
print(f"Positive rate: {y.mean():.4f}")

# ==========================================
# 6. CROSS-VALIDATION (shared folds for A/B/C)
# ==========================================
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

clf = xgb.XGBClassifier(
    n_estimators=150, max_depth=5, learning_rate=0.05,
    subsample=0.8, colsample_bytree=0.8, random_state=42, eval_metric='logloss'
)

def run_cv(feature_list, X_full, y_full, skf, clf):
    """Run 5-fold CV, return AUC/F1/accuracy lists and OOF probabilities."""
    aucs, f1s, accs = [], [], []
    oof_probas = np.zeros(len(y_full))
    for train_idx, test_idx in skf.split(X_full[feature_list], y_full):
        X_tr = X_full.iloc[train_idx][feature_list]
        X_te = X_full.iloc[test_idx][feature_list]
        y_tr, y_te = y_full[train_idx], y_full[test_idx]

        imputer = SimpleImputer(strategy='median')
        X_tr_imp = imputer.fit_transform(X_tr)
        X_te_imp = imputer.transform(X_te)

        clf.fit(X_tr_imp, y_tr)
        p = clf.predict_proba(X_te_imp)[:, 1]
        oof_probas[test_idx] = p

        aucs.append(roc_auc_score(y_te, p))
        f1s.append(f1_score(y_te, (p >= 0.5).astype(int)))
        accs.append(accuracy_score(y_te, (p >= 0.5).astype(int)))
    return aucs, f1s, accs, oof_probas

print("\n" + "=" * 70)
print("RUNNING MODELS A, B, C (5-fold CV)")
print("=" * 70)

# Model A
aucs_a, f1s_a, accs_a, oof_a = run_cv(features_a, model_data, y, skf, clf)
print(f"\nModel A (Shock-only):")
print(f"  AUC: {np.mean(aucs_a):.4f} (±{np.std(aucs_a):.4f})")
print(f"  F1:  {np.mean(f1s_a):.4f}")
print(f"  Acc: {np.mean(accs_a):.4f}")

# Model B
aucs_b, f1s_b, accs_b, oof_b = run_cv(features_b, model_data, y, skf, clf)
print(f"\nModel B (Prior approach, hacohort_c):")
print(f"  AUC: {np.mean(aucs_b):.4f} (±{np.std(aucs_b):.4f})")
print(f"  F1:  {np.mean(f1s_b):.4f}")
print(f"  Acc: {np.mean(accs_b):.4f}")

# Model C
aucs_c, f1s_c, accs_c, oof_c = run_cv(features_c, model_data, y, skf, clf)
print(f"\nModel C (Famine, main effect):")
print(f"  AUC: {np.mean(aucs_c):.4f} (±{np.std(aucs_c):.4f})")
print(f"  F1:  {np.mean(f1s_c):.4f}")
print(f"  Acc: {np.mean(accs_c):.4f}")

# ==========================================
# 7. INCREMENTAL GAINS
# ==========================================
print("\n" + "=" * 70)
print("INCREMENTAL GAINS vs Model A")
print("=" * 70)
print(f"Model B - Model A:  Δ AUC = {np.mean(aucs_b) - np.mean(aucs_a):+.4f}")
print(f"Model C - Model A:  Δ AUC = {np.mean(aucs_c) - np.mean(aucs_a):+.4f}")

# ==========================================
# 8. MODEL D: INTERACTION (famine × hospitalization)
# ==========================================
print("\n" + "=" * 70)
print("MODEL D: Interaction (famine × hospitalization)")
print("=" * 70)

# Build interaction feature
model_data['famine_x_hosp'] = model_data['famine_exposure'] * model_data['r2hosp1y']
features_d = features_c + ['famine_x_hosp']

# Cell sizes for the interaction
print("\nCell sizes (famine × hospitalization):")
cell_table = model_data.groupby(['famine_exposure', 'r2hosp1y']).size().unstack(fill_value=0)
print(cell_table)

print("\nOutcome rate by cell:")
outcome_table = model_data.groupby(['famine_exposure', 'r2hosp1y'])['y_dep_decline_w3'].mean().unstack()
print(outcome_table.round(4))

aucs_d, f1s_d, accs_d, oof_d = run_cv(features_d, model_data, y, skf, clf)
print(f"\nModel D (Famine + interaction):")
print(f"  AUC: {np.mean(aucs_d):.4f} (±{np.std(aucs_d):.4f})")
print(f"  F1:  {np.mean(f1s_d):.4f}")
print(f"  Acc: {np.mean(accs_d):.4f}")
print(f"\nModel D - Model A:  Δ AUC = {np.mean(aucs_d) - np.mean(aucs_a):+.4f}")

# ==========================================
# 9. SUMMARY TABLE
# ==========================================
print("\n" + "=" * 70)
print("SUMMARY: Four-Model Comparison")
print("=" * 70)
summary = pd.DataFrame({
    'Model': ['A (Shock only)', 'B (hacohort_c)', 'C (Famine main)', 'D (Famine × shock)'],
    'Mean AUC': [np.mean(aucs_a), np.mean(aucs_b), np.mean(aucs_c), np.mean(aucs_d)],
    'SD AUC': [np.std(aucs_a), np.std(aucs_b), np.std(aucs_c), np.std(aucs_d)],
    'Mean F1': [np.mean(f1s_a), np.mean(f1s_b), np.mean(f1s_c), np.mean(f1s_d)],
    'Δ AUC vs A': [0.0,
                   np.mean(aucs_b) - np.mean(aucs_a),
                   np.mean(aucs_c) - np.mean(aucs_a),
                   np.mean(aucs_d) - np.mean(aucs_a)]
})
print(summary.round(4).to_string(index=False))

# ==========================================
# 10. SHAP FOR MODEL C (famine feature importance)
# ==========================================
print("\n" + "=" * 70)
print("SHAP FEATURE IMPORTANCE (Model C: Famine)")
print("=" * 70)

X_c_imp = SimpleImputer(strategy='median').fit_transform(model_data[features_c])
clf.fit(X_c_imp, y)
explainer = shap.TreeExplainer(clf)
shap_c = explainer(pd.DataFrame(X_c_imp, columns=features_c))
mean_shap_c = np.abs(shap_c.values).mean(axis=0)

shap_df_c = pd.DataFrame({
    'Feature': features_c,
    'Mean |SHAP|': mean_shap_c
}).sort_values('Mean |SHAP|', ascending=False)
print(shap_df_c.to_string(index=False))

# ==========================================
# 11. SHAP FOR MODEL B (for comparison)
# ==========================================
print("\n" + "=" * 70)
print("SHAP FEATURE IMPORTANCE (Model B: hacohort_c)")
print("=" * 70)

X_b_imp = SimpleImputer(strategy='median').fit_transform(model_data[features_b])
clf.fit(X_b_imp, y)
explainer = shap.TreeExplainer(clf)
shap_b = explainer(pd.DataFrame(X_b_imp, columns=features_b))
mean_shap_b = np.abs(shap_b.values).mean(axis=0)

shap_df_b = pd.DataFrame({
    'Feature': features_b,
    'Mean |SHAP|': mean_shap_b
}).sort_values('Mean |SHAP|', ascending=False)
print(shap_df_b.to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score, f1_score, accuracy_score
from sklearn.impute import SimpleImputer
import shap

# ==========================================
# 1. LOAD & MERGE
# ==========================================
print("Loading CHARLS datasets...")
main_path = r"C:\Users\MIKE\CHARLS H\H_CHARLS_D_Data.dta"
lh_path = r"C:\Users\MIKE\CHARLS_LH\H_CHARLS_LH_a.dta"

df_main = pd.read_stata(main_path)
df_lh = pd.read_stata(lh_path)
df = pd.merge(df_main, df_lh, on='ID', how='inner', suffixes=('', '_lh'))
print(f"Merged shape: {df.shape}")

# ==========================================
# 2. TARGET (with proper NaN handling)
# ==========================================
df['cesd_delta_w2_w3'] = df['r3cesd10'] - df['r2cesd10']
df['y_dep_decline_w3'] = np.where(
    df['r2cesd10'].notna() & df['r3cesd10'].notna(),
    (df['cesd_delta_w2_w3'] > 0).astype(int),
    np.nan
)

# ==========================================
# 3. Famine exposure
# ==========================================
def classify_famine(by):
    if pd.isna(by): return np.nan
    if by >= 1962: return 0
    if 1959 <= by <= 1961: return 1
    if 1956 <= by <= 1958: return 2
    return 3

df['famine_exposure'] = df['rabyear'].apply(classify_famine)

# ==========================================
# 4. HACOHORT numeric
# ==========================================
df['hacohort_c_num'] = pd.to_numeric(
    df['hacohort_c'].cat.codes.replace(-1, np.nan), errors='coerce'
)

# ==========================================
# 5. CONVERT ALL CATEGORICALS (preserve NaN)
# ==========================================
def to_numeric_preserve_nan(s):
    """Convert Stata categorical to numeric, preserving NaN."""
    if str(s.dtype) == 'category':
        # Check if it's a simple yes/no
        vals = set(s.dropna().unique())
        if vals <= {'0.no', '1.yes'}:
            return s.map({'0.no': 0, '1.yes': 1})
        else:
            return pd.to_numeric(s.cat.codes.replace(-1, np.nan), errors='coerce')
    return pd.to_numeric(s, errors='coerce')

# Apply to all relevant variables
for col in ['r2hosp1y', 'r1cesd10', 'r1agey', 'ragender', 'raeduc_c',
            'hacohort_c_num', 'famine_exposure', 'r2cesd10', 'r3cesd10']:
    if col in df.columns:
        df[col] = to_numeric_preserve_nan(df[col])

# ==========================================
# 6. RESTRICT TO ANALYTIC SAMPLE (N = 13,591)
# ==========================================
analytic = df[df['y_dep_decline_w3'].notna()].copy()
print(f"\nAnalytic N: {len(analytic)}")
print(f"Positive rate: {analytic['y_dep_decline_w3'].mean():.4f}")

# ==========================================
# 7. DEFINE FEATURE SETS
# ==========================================
features_a = ['r1cesd10', 'r2hosp1y']
features_age = ['r1cesd10', 'r2hosp1y', 'r1agey']
features_demo = ['r1cesd10', 'r2hosp1y', 'r1agey', 'ragender', 'raeduc_c']
features_b = features_demo + ['hacohort_c_num']
features_c = features_demo + ['famine_exposure']

# Add missingness indicator for r2hosp1y
analytic['r2hosp1y_missing'] = analytic['r2hosp1y'].isna().astype(int)
for fs in [features_a, features_age, features_demo, features_b, features_c]:
    fs.append('r2hosp1y_missing')

y = analytic['y_dep_decline_w3'].values

# ==========================================
# 8. CROSS-VALIDATION
# ==========================================
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

clf = xgb.XGBClassifier(
    n_estimators=150, max_depth=5, learning_rate=0.05,
    subsample=0.8, colsample_bytree=0.8, random_state=42, eval_metric='logloss'
)

def run_cv(feature_list, X_full, y_full, skf, clf):
    aucs, f1s, accs = [], [], []
    oof_probas = np.zeros(len(y_full))
    for train_idx, test_idx in skf.split(X_full[feature_list], y_full):
        X_tr = X_full.iloc[train_idx][feature_list]
        X_te = X_full.iloc[test_idx][feature_list]
        y_tr, y_te = y_full[train_idx], y_full[test_idx]

        imputer = SimpleImputer(strategy='median')
        X_tr_imp = imputer.fit_transform(X_tr)
        X_te_imp = imputer.transform(X_te)

        clf.fit(X_tr_imp, y_tr)
        p = clf.predict_proba(X_te_imp)[:, 1]
        oof_probas[test_idx] = p

        aucs.append(roc_auc_score(y_te, p))
        f1s.append(f1_score(y_te, (p >= 0.5).astype(int)))
        accs.append(accuracy_score(y_te, (p >= 0.5).astype(int)))
    return aucs, f1s, accs, oof_probas

# ==========================================
# 9. RUN ALL FIVE MODELS
# ==========================================
print("\n" + "=" * 70)
print("RUNNING FIVE MODELS ON CORRECTED SAMPLE")
print("=" * 70)

models = {
    'A (Shock only)': features_a,
    'A + Age': features_age,
    'A + Demographics': features_demo,
    'B (hacohort_c)': features_b,
    'C (Famine)': features_c,
}

results = {}
for name, feats in models.items():
    aucs, f1s, accs, oof = run_cv(feats, analytic, y, skf, clf)
    results[name] = {
        'aucs': aucs, 'f1s': f1s, 'accs': accs, 'oof': oof,
        'mean_auc': np.mean(aucs),
        'std_auc': np.std(aucs),
        'mean_f1': np.mean(f1s),
        'mean_acc': np.mean(accs),
    }
    print(f"\n{name}:")
    print(f"  AUC: {np.mean(aucs):.4f} (±{np.std(aucs):.4f})")
    print(f"  F1:  {np.mean(f1s):.4f}")
    print(f"  Acc: {np.mean(accs):.4f}")

# ==========================================
# 10. SUMMARY TABLE
# ==========================================
print("\n" + "=" * 70)
print("FINAL SUMMARY TABLE")
print("=" * 70)

baseline_auc = results['A (Shock only)']['mean_auc']
summary_rows = []
for name, r in results.items():
    summary_rows.append({
        'Model': name,
        'Mean AUC': f"{r['mean_auc']:.4f}",
        'SD': f"±{r['std_auc']:.4f}",
        'Mean F1': f"{r['mean_f1']:.4f}",
        'Δ AUC vs A': f"{r['mean_auc'] - baseline_auc:+.4f}",
    })

summary_df = pd.DataFrame(summary_rows)
print(summary_df.to_string(index=False))
summary_df.to_csv('table2_model_comparison.csv', index=False)

# ==========================================
# 11. DECOMPOSITION: HOW MUCH IS LIFE-COURSE?
# ==========================================
print("\n" + "=" * 70)
print("GAIN DECOMPOSITION")
print("=" * 70)

gain_age = results['A + Age']['mean_auc'] - baseline_auc
gain_demo = results['A + Demographics']['mean_auc'] - baseline_auc
gain_b = results['B (hacohort_c)']['mean_auc'] - baseline_auc
gain_c = results['C (Famine)']['mean_auc'] - baseline_auc

print(f"\nModel A (baseline):              {baseline_auc:.4f}")
print(f"Model A + Age:                   Δ = {gain_age:+.4f}")
print(f"Model A + Demographics:          Δ = {gain_demo:+.4f}")
print(f"Model B (hacohort_c):            Δ = {gain_b:+.4f}")
print(f"Model C (Famine):                Δ = {gain_c:+.4f}")

print(f"\nResidual life-course contribution:")
print(f"  B (hacohort_c beyond demographics):  {gain_b - gain_demo:+.4f}")
print(f"  C (famine beyond demographics):      {gain_c - gain_demo:+.4f}")

# ==========================================
# 12. SHAP FOR MODELS B AND C
# ==========================================
print("\n" + "=" * 70)
print("SHAP: MODEL B (hacohort_c)")
print("=" * 70)

X_b_imp = SimpleImputer(strategy='median').fit_transform(analytic[features_b])
clf.fit(X_b_imp, y)
explainer = shap.TreeExplainer(clf)
shap_b = explainer(pd.DataFrame(X_b_imp, columns=features_b))
mean_shap_b = np.abs(shap_b.values).mean(axis=0)

shap_df_b = pd.DataFrame({
    'Feature': features_b,
    'Mean |SHAP|': mean_shap_b
}).sort_values('Mean |SHAP|', ascending=False)
print(shap_df_b.to_string(index=False))
shap_df_b.to_csv('table4_shap_model_b.csv', index=False)

print("\n" + "=" * 70)
print("SHAP: MODEL C (Famine)")
print("=" * 70)

X_c_imp = SimpleImputer(strategy='median').fit_transform(analytic[features_c])
clf.fit(X_c_imp, y)
explainer = shap.TreeExplainer(clf)
shap_c = explainer(pd.DataFrame(X_c_imp, columns=features_c))
mean_shap_c = np.abs(shap_c.values).mean(axis=0)

shap_df_c = pd.DataFrame({
    'Feature': features_c,
    'Mean |SHAP|': mean_shap_c
}).sort_values('Mean |SHAP|', ascending=False)
print(shap_df_c.to_string(index=False))
shap_df_c.to_csv('table4_shap_model_c.csv', index=False)
```

```python
import pandas as pd
import numpy as np
from scipy import stats
import statsmodels.api as sm

main_path = r"C:\Users\MIKE\CHARLS H\H_CHARLS_D_Data.dta"
lh_path = r"C:\Users\MIKE\CHARLS_LH\H_CHARLS_LH_a.dta"

df_main = pd.read_stata(main_path)
df_lh = pd.read_stata(lh_path)
df = pd.merge(df_main, df_lh, on='ID', how='inner', suffixes=('', '_lh'))

# --- Force numeric conversions with float type everywhere ---
def to_float(s):
    if str(s.dtype) == 'category':
        vals = set(s.dropna().unique())
        if vals <= {'0.no', '1.yes'}:
            return s.map({'0.no': 0.0, '1.yes': 1.0}).astype(float)
        if vals <= {'1.man', '2.woman'}:
            return s.map({'1.man': 1.0, '2.woman': 2.0}).astype(float)
        return pd.to_numeric(s.cat.codes.replace(-1, np.nan), errors='coerce').astype(float)
    return pd.to_numeric(s, errors='coerce').astype(float)

df['hosp'] = to_float(df['r2hosp1y'])
df['cesd_w2'] = to_float(df['r2cesd10'])
df['cesd_w3'] = to_float(df['r3cesd10'])
df['age'] = to_float(df['r1agey'])
df['gender'] = to_float(df['ragender'])
df['cesd_delta'] = df['cesd_w3'] - df['cesd_w2']

valid = df[df['cesd_w2'].notna() & df['cesd_w3'].notna() & df['hosp'].notna()].copy()
print(f"Analytic N: {len(valid)}")
print(f"hosp dtype: {valid['hosp'].dtype}")
print(f"cesd_w2 dtype: {valid['cesd_w2'].dtype}")

# Standardize
valid['cesd_w2_std'] = (valid['cesd_w2'] - valid['cesd_w2'].mean()) / valid['cesd_w2'].std()
valid['age_std'] = (valid['age'] - valid['age'].mean()) / valid['age'].std()

# Interaction
valid['hosp_x_cesd'] = valid['hosp'].astype(float) * valid['cesd_w2_std'].astype(float)

print("\n" + "=" * 70)
print("FORMAL INTERACTION TEST: Hospitalization × Baseline CES-D")
print("=" * 70)

X = valid[['hosp', 'cesd_w2_std', 'hosp_x_cesd', 'age_std', 'gender']].dropna()
X = sm.add_constant(X)
y = valid.loc[X.index, 'cesd_delta']
model = sm.OLS(y, X).fit()
print(model.summary())

print("\n" + "=" * 70)
print("INTERACTION COEFFICIENT")
print("=" * 70)
print(f"hosp_x_cesd:  {model.params['hosp_x_cesd']:+.4f}")
print(f"  SE:         {model.bse['hosp_x_cesd']:.4f}")
print(f"  p-value:    {model.pvalues['hosp_x_cesd']:.4f}")
print(f"  95% CI:     [{model.conf_int().loc['hosp_x_cesd', 0]:+.4f}, {model.conf_int().loc['hosp_x_cesd', 1]:+.4f}]")

print(f"\nInterpretation:")
if model.pvalues['hosp_x_cesd'] < 0.05:
    print(f"  ✓ Significant interaction (p < 0.05)")
    print(f"  → Baseline CES-D significantly modifies the hospitalization effect")
else:
    print(f"  ✗ Interaction not significant")

# Sensitivity: alternate outcomes
print("\n" + "=" * 70)
print("SENSITIVITY: Alternate Binary Outcomes with Interaction")
print("=" * 70)
from statsmodels.formula.api import logit

valid['outcome_2pt'] = (valid['cesd_delta'] >= 2).astype(int)
valid['outcome_4pt'] = (valid['cesd_delta'] >= 4).astype(int)
valid['outcome_any'] = (valid['cesd_delta'] > 0).astype(int)

for outcome_name in ['outcome_any', 'outcome_2pt', 'outcome_4pt']:
    formula = f"{outcome_name} ~ hosp * cesd_w2_std + age_std + gender"
    logit_model = logit(formula, data=valid).fit(disp=0)
    interaction_p = logit_model.pvalues['hosp:cesd_w2_std']
    interaction_coef = logit_model.params['hosp:cesd_w2_std']
    print(f"\n{outcome_name}:")
    print(f"  Interaction coef: {interaction_coef:+.4f}, p = {interaction_p:.4f}")
    if interaction_p < 0.05:
        print(f"  → Significant interaction")
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

main_path = r"C:\Users\MIKE\CHARLS H\H_CHARLS_D_Data.dta"
lh_path = r"C:\Users\MIKE\CHARLS_LH\H_CHARLS_LH_a.dta"

df_main = pd.read_stata(main_path)
df_lh = pd.read_stata(lh_path)
df = pd.merge(df_main, df_lh, on='ID', how='inner', suffixes=('', '_lh'))

def to_float(s):
    if str(s.dtype) == 'category':
        vals = set(s.dropna().unique())
        if vals <= {'0.no', '1.yes'}:
            return s.map({'0.no': 0.0, '1.yes': 1.0}).astype(float)
        if vals <= {'1.man', '2.woman'}:
            return s.map({'1.man': 1.0, '2.woman': 2.0}).astype(float)
        return pd.to_numeric(s.cat.codes.replace(-1, np.nan), errors='coerce').astype(float)
    return pd.to_numeric(s, errors='coerce').astype(float)

df['hosp'] = to_float(df['r2hosp1y'])
df['cesd_w1'] = to_float(df['r1cesd10'])
df['cesd_w2'] = to_float(df['r2cesd10'])
df['cesd_w3'] = to_float(df['r3cesd10'])
df['age'] = to_float(df['r1agey'])
df['gender'] = to_float(df['ragender'])
df['cesd_delta'] = df['cesd_w3'] - df['cesd_w2']

# Keep ONLY complete cases for the full model
required = ['hosp', 'cesd_w2', 'cesd_w3', 'cesd_delta', 'age', 'gender']
valid = df.dropna(subset=required).copy()
print(f"Analytic N: {len(valid)}")

print("=" * 70)
print("FINAL DIAGNOSTIC: Is the hospitalization effect robust?")
print("=" * 70)

# 1. Unadjusted
X1 = sm.add_constant(valid[['hosp']])
m1 = sm.OLS(valid['cesd_delta'], X1).fit()
print(f"\n1. Unadjusted:                          hosp coef = {m1.params['hosp']:+.4f}, p = {m1.pvalues['hosp']:.4f}")

# 2. Adjusted for baseline CES-D
X2 = sm.add_constant(valid[['hosp', 'cesd_w2']])
m2 = sm.OLS(valid['cesd_delta'], X2).fit()
print(f"2. Adjusted for baseline CES-D:         hosp coef = {m2.params['hosp']:+.4f}, p = {m2.pvalues['hosp']:.4f}")

# 3. Adjusted for CES-D + age + gender
X3 = sm.add_constant(valid[['hosp', 'cesd_w2', 'age', 'gender']])
m3 = sm.OLS(valid['cesd_delta'], X3).fit()
print(f"3. Adjusted + age + gender:             hosp coef = {m3.params['hosp']:+.4f}, p = {m3.pvalues['hosp']:.4f}")

# 4. Trimmed (5-95% baseline CES-D)
q05 = valid['cesd_w2'].quantile(0.05)
q95 = valid['cesd_w2'].quantile(0.95)
trim = valid[(valid['cesd_w2'] >= q05) & (valid['cesd_w2'] <= q95)]
X4 = sm.add_constant(trim[['hosp', 'cesd_w2', 'age', 'gender']])
m4 = sm.OLS(trim['cesd_delta'], X4).fit()
print(f"4. Trimmed (5-95% baseline):            hosp coef = {m4.params['hosp']:+.4f}, p = {m4.pvalues['hosp']:.4f} (N={len(trim)})")

# 5. Adjusted for W1 CES-D too
has_w1 = valid.dropna(subset=['cesd_w1'])
X5 = sm.add_constant(has_w1[['hosp', 'cesd_w2', 'cesd_w1', 'age', 'gender']])
m5 = sm.OLS(has_w1['cesd_delta'], X5).fit()
print(f"5. Adjusted + W1 CES-D:                 hosp coef = {m5.params['hosp']:+.4f}, p = {m5.pvalues['hosp']:.4f} (N={len(has_w1)})")

# 6. Trimming extreme deltas
normal_range = valid[valid['cesd_delta'].abs() <= 20]
X6 = sm.add_constant(normal_range[['hosp', 'cesd_w2', 'age', 'gender']])
m6 = sm.OLS(normal_range['cesd_delta'], X6).fit()
print(f"6. Trimming extreme deltas (|Δ|≤20):    hosp coef = {m6.params['hosp']:+.4f}, p = {m6.pvalues['hosp']:.4f} (N={len(normal_range)})")

# 7. Multiple imputation-style: mode of the coefficients
print("\n" + "=" * 70)
print("SUMMARY OF HOSPITALIZATION EFFECT ACROSS SPECIFICATIONS")
print("=" * 70)
coefs = [
    ('Unadjusted', m1.params['hosp'], m1.pvalues['hosp']),
    ('+ Baseline CES-D', m2.params['hosp'], m2.pvalues['hosp']),
    ('+ Demographics', m3.params['hosp'], m3.pvalues['hosp']),
    ('+ Trimmed baseline', m4.params['hosp'], m4.pvalues['hosp']),
    ('+ W1 CES-D', m5.params['hosp'], m5.pvalues['hosp']),
    ('+ Trimmed deltas', m6.params['hosp'], m6.pvalues['hosp']),
]
print(f"\n{'Specification':<30} {'Coef':>10} {'p-value':>10}")
print("-" * 70)
for name, coef, p in coefs:
    print(f"{name:<30} {coef:>+10.4f} {p:>10.4f}")

print("\n" + "=" * 70)
print("INTERPRETATION")
print("=" * 70)
print(f"\nUnadjusted effect: {m1.params['hosp']:+.4f} (p = {m1.pvalues['hosp']:.4f})")
print(f"Adjusted effect:   {m2.params['hosp']:+.4f} (p = {m2.pvalues['hosp']:.4f})")
print(f"\n→ The apparent null in the unadjusted analysis is due to")
print(f"  confounding by baseline depression level.")
print(f"→ After adjustment, hospitalization predicts a")
print(f"  {m2.params['hosp']:+.2f}-point increase in CES-D change.")
```
